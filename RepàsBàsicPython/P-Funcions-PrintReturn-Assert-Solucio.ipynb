**1**. Escriu una funció que rebi dos nombres com a paràmetres i retorni la seva suma.

```python

def sumar(primer_sumand: int, segon_sumand: int) -> int:
    return primer_sumand + segon_sumand
```

```python
# Exemple d'ús:
resultat: int = sumar(5, 7)
print(resultat)

assert sumar(5, 7) == 12
```

```python
# Aquest assert falla intencionadament: 5 + 7 és 12, no 10.
# Serveix per observar com assert detecta una expectativa incorrecta.
assert sumar(5, 7) == 10
```

**2**. Escriu una funció _resta(a,b)_ que resti dos nombres i retorni el resultat.

```python

def restar(minuend: int, subtrahend: int) -> int:
    return minuend - subtrahend
```

**3**. Escriu una funció _multiplica(a,b)_ que multipliqui dos nombres i retorni el resultat.

```python

def multiplicar(primer_factor: int, segon_factor: int) -> int:
    return primer_factor * segon_factor
```

**4**. Escriu una funció _divideix(a,b)_ que divideixi dos nombres i retorni el resultat.

```python

def dividir(dividend: int, divisor: int) -> float:
    return dividend / divisor
```

**5**. Escriu una funció _area_cercle(radi)_ que calculi l'àrea d'un cercle donat el seu radi.

Si et cal importa la llibreria math i usa alguna de les seves constants.

```python

import math


def calcular_area_cercle(radi: float) -> float:
    return math.pi * radi**2
```

**6**. Escriu una funció _celsius_a_fahrenheit(celsius)_  que converteixi graus Celsius a Fahrenheit.

```python

def convertir_celsius_a_fahrenheit(celsius: float) -> float:
    return celsius * 9 / 5 + 32
```

**7**. Escriu una funció que converteixi graus Fahrenheit a Celsius.

```python

def convertir_fahrenheit_a_celsius(fahrenheit: float) -> float:
    return (fahrenheit - 32) * 5 / 9
```

**8**.  Escriu una funció que calculi la mitjana de tres números.

```python

def calcular_mitjana(primer_nombre: float, segon_nombre: float, tercer_nombre: float) -> float:
    return (primer_nombre + segon_nombre + tercer_nombre) / 3
```

**9**. Escriu una funció que determini si un nombre és parell o senar.

```python

def es_parell(nombre: int) -> bool:
    return nombre % 2 == 0
```

```python
# Exemple d'ús:
resultat: bool = es_parell(7)
print(resultat)
```

**10**. Escriu una funció que retorni el valor absolut d'un nombre.

```python

def calcular_valor_absolut(nombre: int) -> int:
    if nombre < 0:
        return -nombre
    return nombre


# També existeix la funció incorporada abs().
```
