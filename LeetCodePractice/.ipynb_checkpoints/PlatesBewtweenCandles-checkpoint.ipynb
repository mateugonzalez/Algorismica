```python
def platesBetweenCandles(s: str, queries: list[list[int]]) -> list[int]:
    list1 = []
    list2 = []
    boxes = list(s.split("|"))
    for i in range(1, len(boxes)-1):
        list1.append(boxes[i])
    for t in range(len(list1)):
        h = int(len(list1[t]))
        list2.append(h)
    return list2

        

platesBetweenCandles("***|**|*****|**||**|*", [1,3])
            
            
                    
            
            
```
