```python
def findMedianSortedArrays(nums1: list[int], nums2: list[int]) -> float:
    nums3 = nums1 + nums2
    nums3.sort()
    n = len(nums3)
    if n % 2 == 0:
        output = float((nums3[(n//2)-1] + nums3[n//2])/2)
    else:
        output = float(nums3[n//2])
        
    return output

findMedianSortedArrays([1, 2], [3,8, 8, 0, 4])
```
