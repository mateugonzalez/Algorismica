```python
def fourSum(nums: list[int], target: int) -> list[list[int]]:
    result = []
    n = len(nums)
    for i in range(n):
        for t in range(i+1, n):
            for y in range(t+1, n):
                for u in range(y+1, n):
                    if nums[i] + nums[t] + nums[y] + nums[u] == target:
                        quadruplet = [nums[i], nums[t], nums[y], nums[u]]
                        result.append(quadruplet)
    return result
fourSum([1,0,-1,0,-2,2], 0)         


            
                
```
