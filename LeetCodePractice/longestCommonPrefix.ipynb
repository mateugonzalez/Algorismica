```python
def longestCommonPrefix(strs: list[str]) -> str:
    firststr: list = list(strs[0])            #O(n+1)
    max_count: int = len(firststr)            #O(n)
    for i in range(1, len(strs)): 
        word: list = list(strs[i])
        print(word)
        counter: int = 0
        n = len(firststr)
        m = len(word)
        if m > n:
            for h in range(n):
                if word[h] == firststr[h]:
                    counter +=1
            if counter < max_count:
                max_count = counter
        else:
            for k in range(m):
                if word[k] == firststr[k]:
                    counter +=1
            if counter < max_count:
                max_count = counter
    return max_count
            
                
        
longestCommonPrefix(["flowers", "flow"])       
            
            
```

```python
import timeit
start_time = timeit.default_timer()
longestCommonPrefix(["flowers", "flow"])
print(timeit.default_timer() - start_time)
```
