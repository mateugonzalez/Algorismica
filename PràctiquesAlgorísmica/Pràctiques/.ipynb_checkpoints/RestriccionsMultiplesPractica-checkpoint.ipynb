```python
import math
def maxim_restringit() -> int:
    k = 1
    list1 = []
    for i in range(1, 12499):
        k *= i
        n = (4 * k) + 4
        u = math.isqrt(n)
        if u * u == n:
            list1.append(n)
        else:
            pass
    return list1[-1]

maxim_restringit()
```

```python
assert maxim_restringit() == 20164
```

```python
def blocs_factorials() -> list[int]:
    l = 1
    list2 = []
    for t in range(1, 11):
        l *= t
        b = l//720
        if l % 720 == 0:
            list2.append(b)
        else:
            pass
    return list2
blocs_factorials()
```

```python
assert blocs_factorials() == [1, 7, 56, 504, 5040]
```
