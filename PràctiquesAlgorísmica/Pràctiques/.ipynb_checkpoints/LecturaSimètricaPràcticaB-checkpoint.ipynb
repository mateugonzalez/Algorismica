```python
def primera_rotacio_simetrica(codi: str) -> str:
    """Retorna la primera rotació simètrica o ''."""
    list1 = list(codi)
    n = len(list1)
    simetry = ""
    for t in range(n):
        list2 = list1[t:] + list1[:t]
        if list2[0:(len(list2)//2)] == list2[-1:(len(list2)//2 -1 + len(list2) % 2):-1]:
            result = "".join(list2)
            return result
        else:
            continue
    return simetry
    
    
        
    

primera_rotacio_simetrica("BCBCA")
        
            
        
```
