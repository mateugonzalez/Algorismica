```python
def indexs_simetria(valors: list[int], centre: int) -> list[tuple[int, int]]:
    """Retorna parelles d'índexs originals amb valors simètrics."""
    llistasp = []
    usats = []
    for i in valors:
        if centre > i:
            increment_a = centre - i
            j= centre + increment_a
            if j in valors:
                k = valors.index(i)
                g = valors.index(j)
                if k<g and k not in usats and g not in usats:
                    llistasp.append((k,g))
                    usats.append(k)
                    usats.append(g)
                else:
                    llistasp.append((g,k))
        else:
            pass
    llistasp.sort()
    return llistasp
indexs_simetria([14, 2, 9, 6, 18, 10], 10)
```

```python
assert indexs_simetria([14, 2, 9, 6, 18, 10], 10) == [(0, 3), (1, 4)]
```
