```python
def base12_a_decimal(xifra: str) -> int:
    """
    Aquesta funció transforma un nombre 
    en base 12 a nombre decimal.

    Parameters
    ----------
    xifra: str

    Returns
    -------
    decimal: int      
    """    
    list1 = list(xifra)
    n = len(list1)
    result = 0
    for i in range(n):
        pot = n - i -1
        mult = 12 ** pot
        if list1[i] == "A":
            result += mult * 10
        elif list1[i] == "B":
            result += mult * 11
        else:
            num = int(list1[i])
            result += mult * num
    return result
base12_a_decimal("A2")
```

```python
assert base12_a_decimal('2A') == 34
assert base12_a_decimal('B3') == 135
```

```python
def nombre_digits(valor: int) -> int:
    """
    Aquesta funció retorna el nombre de digits 
    d'un nombre natural sense convertir-lo 
    a cadena.

    Parameters
    ----------
    valor: int
        Un nombre natural.

    Returns
    -------
    numDigits: int
        Quantitat de digits que té valor.
    """   
    dig_dec = 0
    while valor > 0:
        valor = valor // 10
        dig_dec += 1
    return dig_dec
nombre_digits(187763)
```
