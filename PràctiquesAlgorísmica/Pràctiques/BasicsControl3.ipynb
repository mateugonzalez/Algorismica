```python
def narcisista(n: int) -> bool:
    """
    Comprova si un nombre és narcisista.

    Un nombre és narcisista si és igual a la suma dels seus dígits
    elevats a la potència del nombre de dígits.

    Parameters
    ----------
    n : int
        Nombre enter que es vol comprovar.

    Returns
    -------
    resultat : bool
        True si el nombre és narcisista; False en cas contrari.
    """
    llista1 = list(str(n))
    potencia = int(len(llista1))
    righteq = 0
    for i in range(len(llista1)):
        righteq += int(llista1[i])**(potencia)
    if righteq == n:
        return True
    else:
        return False
    

narcisista(9)
```
