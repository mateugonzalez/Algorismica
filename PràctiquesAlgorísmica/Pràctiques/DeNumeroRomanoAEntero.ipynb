```python
def romanToInt(s: str) -> int:
    total = 0
    if s.isupper():
        valores = {
            'I': 1,
            'V': 5,
            'X': 10,
            'L': 50,
            'C': 100,
            'D': 500,
            'M': 1000
        }
        for i in range(len(s)):
            if i+2 in range(len(s)):
                if valores[s[i]] < valores[s[i+1]] and valores[s[i+1]] < valores[s[i+2]]:
                    print("Syntax Error")
                    return None
                else:
                    pass
            if i+1 in range(len(s)):
                if valores[s[i]] >= valores[s[i+1]]:
                    total += valores[s[i]]
                else:
                    total -= valores[s[i]]
            else:
                total += valores[s[i]]
    else:
        print("Syntax Error")
    return total
result = romanToInt("MMMCMXCIX")
print(result)
```
