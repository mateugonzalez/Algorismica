```python
def productes_acumulats(lim: int) -> list[int]:
    """Retorna els termes de la successió estrictament menors que lim."""
    if lim <= 1:
        return []
    
    number_list = [1]
    succesion_entero = 1
    
    while number_list[-1] < lim:
        succesion_entero = succesion_entero +1
        seguent = number_list[-1]*(succesion_entero)
        if seguent < lim:
            number_list.append(seguent)
        else:
            break
    
    return number_list

assert productes_acumulats(25) == [1, 2, 6, 24]
assert productes_acumulats(7) == [1, 2, 6]
assert productes_acumulats(2) == [1]

productes_acumulats(70)
```
