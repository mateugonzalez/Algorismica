```python
def min_max() -> None:
    """
    Demana nombres enters a l'usuari i mostra el mínim i el màxim introduïts.

    La introducció de -99 indica que s'ha acabat la seqüència de nombres.

    Parameters
    ----------
    No té paràmetres.

    Returns
    -------
    None
        La funció mostra el mínim i el màxim per pantalla.
    """
    number1 = int(input())
    list1 = []
    while number1 != -99:
        list1.append(number1)
        number1 = int(input())
    number2 = -99
    list1.append(number2)
    list1.sort()
    print(list1[0], list1[-1])
min_max()
```
