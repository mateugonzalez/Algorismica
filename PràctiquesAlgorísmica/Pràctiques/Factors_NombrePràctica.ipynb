```python
def factors_primers(n) -> list[int]:
    p = 2
    list1 = []
    while n > 1:
        t = n / p
        if n % p == 0:
            list1.append(p)
            n = t
        else:
            p += 1
    return list1

assert factors_primers(600851475143) == [71, 839, 1471, 6857]
assert factors_primers(12) == [2, 2, 3]
```
