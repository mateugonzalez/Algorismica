```python
def removeElement(nums: list[int], val: int) -> int:
    list1 = []
    list2 = []
    for i in nums:
        if val == i:
            list2.append("_")
        else:
            list1.append(i)
    list3 = list1 + list2
    return list3
result = removeElement([3,2,2,3], 3)
print(result)
```
