```python
def codi_simmetric(codi: str) -> bool:
    list1 = list(codi)
    count = 0
    for i in range(len(list1)):
        if list1[count] == list1[len(list1)-1-count]:
            count += 1
            continue
        else:
            return False
    return True
            
```

```python
codi_simmetric("ABBBBBBBABBBBA")
```

```python
assert codi_simmetric('ABCCBA') is True
assert codi_simmetric('ABCA') is False
assert codi_simmetric('X') is True
```
