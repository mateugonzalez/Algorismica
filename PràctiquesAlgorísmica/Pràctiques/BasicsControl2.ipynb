```python
import random
def aleatori(s: str, n: int) -> str:
    """
    Genera una cadena aleatòria a partir dels caràcters d'una cadena donada.

    Parameters
    ----------
    s : str
        Cadena que conté els caràcters disponibles.
    n : int
        Longitud de la cadena que es vol generar.

    Returns
    -------
    resultat : str
        Cadena de longitud n formada per caràcters seleccionats aleatòriament.
    """
    t = ""
    k = list(s)
    for l in range(n):
        l = str(random.choice(k))
        t += l
    result = t
    return result
print(aleatori("maria", 8))
        
        


```
