```python
def valida_codi_modular(codi: str) -> bool:
    list1 = list(codi)
    lletra = list1[-1]
    list1 = list1[:-1]
    num = int("".join(list1))
    u = "QWERTYUIOPASDFGHJ"
    y = list(u)
    i = num % 17
    if lletra == y[i]:
        return True
    else:
        return False
```

```python
assert valida_codi_modular('12345R') is True
assert valida_codi_modular('12345D') is False
assert valida_codi_modular('34Q') is True
```
